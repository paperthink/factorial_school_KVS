```python
# factorial of  n
#n x n-1 x n-2, where n-i, i+=1
```

```python
i = 1


n = int(input("Enter a number: "))
p = 1
while i <= n:
    p*=i
    i+=1

print(f"factorial of the number is {p}")
```
